```python
# 1. 무료 AI 라이브러리 설치 (Shift+Enter 로 실행)
!pip install diffusers transformers accelerate

# 2. 코랩 무료 GPU에 AI 모델 탑재하기 (API 키 불필요!)
from diffusers import StableDiffusionPipeline
import torch

print("AI 모델을 코랩으로 가져오는 중입니다... (약 1분 소요)")
pipe = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16)
pipe = pipe.to("cuda") # 코랩의 무료 그래픽카드 사용

# 3. 나만의 프롬프트(명령어)로 그림 생성하기
prompt = "나는 나는 천사야"
image = pipe(prompt).images[0]

# 4. 결과물 확인
image
```

### GPU로 런타임 바꿔서 다시 해보기

```python
# 1. 무료 AI 라이브러리 설치
!pip install diffusers transformers accelerate

# 2. AI 모델 탑재 (검증된 모델로 수정)
from diffusers import StableDiffusionPipeline
import torch

print("AI 모델을 가져오는 중입니다... (약 1분 소요)")

# 주소가 확실한 공식 모델을 사용합니다.
model_id = "runwayml/stable-diffusion-v1-5"
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)
pipe = pipe.to("cuda")

# 3. 그림 생성
prompt = "나는 천사야, anime style, high quality, detailed"
image = pipe(prompt).images[0]

# 4. 결과물 확인
image
```

```python
# 뉴스 제목 수집기 만들기 (API 불필요, 100% 무료)
import requests
from bs4 import BeautifulSoup

url = "https://news.naver.com/"
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

print("=== 오늘 네이버 뉴스 헤드라인 ===")
for headline in soup.select('.cjs_t')[:5]: # 5개만 가져오기
    print("📰", headline.text)
```
